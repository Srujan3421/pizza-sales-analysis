```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('pizza_sales.csv')
```

```python

numeric_df = df.select_dtypes(include=np.number)
corr_matrix = numeric_df.corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title("Correlation Matrix of Numeric Features")
plt.show()
```

```python
# List all columns
print(df.columns)

# Check unique values in categorical columns
for col in df.select_dtypes(include='object').columns:
    print(col,df[col].unique())
```

```python
# Total sales
if 'total_price' in df.columns:
    print('Total Sales:', df['total_price'].sum())
```

```python
plt.figure(figsize=(10, 6))
sales_by_category = df.groupby('pizza_category')['total_price'].sum().sort_values(ascending=False)
sns.barplot(x=sales_by_category.index, y=sales_by_category.values)
plt.title('Total Sales by Pizza Category')
plt.xlabel('Pizza Category')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Create second plot - Top 10 Best-Selling Pizzas
plt.figure(figsize=(12, 6))
top_pizzas = df.groupby('pizza_name')['quantity'].sum().sort_values(ascending=False).head(10)
sns.barplot(x=top_pizzas.index, y=top_pizzas.values)
plt.title('Top 10 Best-Selling Pizzas')
plt.xlabel('Pizza Name')
plt.ylabel('Quantity Sold')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Pizza size distribution
plt.figure(figsize=(10, 6))
df['pizza_size'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Distribution of Pizza Sizes')
plt.axis('equal')
plt.show()

# Average price by size and category
avg_price_size_cat = df.groupby(['pizza_size', 'pizza_category'])['unit_price'].mean().unstack()
plt.figure(figsize=(12, 6))
avg_price_size_cat.plot(kind='bar')
plt.title('Average Pizza Price by Size and Category')
plt.xlabel('Pizza Size')
plt.ylabel('Average Price ($)')
plt.legend(title='Pizza Category', bbox_to_anchor=(1.05, 1))
plt.tight_layout()
plt.show()
```

```python
# 1. Histogram of Pizza Prices
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x='total_price', bins=30)
plt.title('Distribution of Pizza Prices')
plt.xlabel('Price ($)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()

# 2. Box Plot of Prices by Category
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='pizza_category', y='total_price')
plt.title('Price Distribution by Pizza Category')
plt.xlabel('Category')
plt.ylabel('Price ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. Box Plot of Prices by Size
plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x='pizza_size', y='total_price')
plt.title('Price Distribution by Pizza Size')
plt.xlabel('Size')
plt.ylabel('Price ($)')
plt.tight_layout()
plt.show()

# 4. Scatter Plot of Quantity vs Total Price
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='quantity', y='total_price', alpha=0.5)
plt.title('Relationship between Quantity and Total Price')
plt.xlabel('Quantity Ordered')
plt.ylabel('Total Price ($)')
plt.tight_layout()
plt.show()

# 5. Box Plot of Quantity by Category
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='pizza_category', y='quantity')
plt.title('Order Quantity Distribution by Pizza Category')
plt.xlabel('Category')
plt.ylabel('Quantity')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 6. Violin Plot of Prices by Category
plt.figure(figsize=(12, 6))
sns.violinplot(data=df, x='pizza_category', y='total_price')
plt.title('Price Distribution (Violin Plot) by Pizza Category')
plt.xlabel('Category')
plt.ylabel('Price ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 7. Create a correlation heatmap for numerical columns
numeric_cols = df.select_dtypes(include=[np.number]).columns
plt.figure(figsize=(10, 8))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap of Numerical Variables')
plt.tight_layout()
plt.show()
```
